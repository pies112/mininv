# 4. Финальная обработка патентной базы 2020–2025

Ноутбук берет готовый файл `3_patents_2020_2025_final.xlsx`, при необходимости обращается к шести исходным CSV Роспатента и формирует финальную версию 4.

Для всех типов объектов добавляются страна регистрации, страны авторов и страны правообладателей. Полностью удаляется объект со статусом `parse_validation_failed`. Промышленные образцы сохраняются с `RU`, международностью `0` и цитируемостью `0`.

Запустите **Run All**.

```python
from pathlib import Path
from urllib.request import urlopen
import runpy

SCRIPT_URL = 'https://raw.githubusercontent.com/pies112/mininv/main/chatgpt_artifacts/4_patent_finalizer.py'
SCRIPT_PATH = Path.cwd() / '4_patent_finalizer.py'

print('Загрузка исполняемого модуля версии 4...')
SCRIPT_PATH.write_bytes(urlopen(SCRIPT_URL, timeout=60).read())
print('Сохранено:', SCRIPT_PATH)

result = runpy.run_path(str(SCRIPT_PATH), run_name='__main__')
outputs = result.get('outputs')
outputs
```
